Pandas DataFrame :- A two-dimensional tabular structure containing rows and columns.

```python
import numpy as np
import pandas as pd
```

Creating a dataframe

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]

df2 = pd.DataFrame(data_list)
```

```python
coloumns = ["Name", "Age", "City", "Salary"]
df2 = pd.DataFrame(data_list, columns = coloumns)
```
