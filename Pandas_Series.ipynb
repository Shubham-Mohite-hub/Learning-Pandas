Pandas Series :- A one-dimensional labeled array, similar to a single column.

```python
There are multiple ways to create pandas series
```

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
my_list = [10, 20, 30]
my_array = np.array([10, 20, 30])
d = {'a':10, 'b':20, 'c': 30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(my_array)
```

```python
pd.Series(d)
```
